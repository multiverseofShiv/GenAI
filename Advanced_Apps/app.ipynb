```python
import os
from dotenv import load_dotenv
from langchain_groq import ChatGroq


load_dotenv()

model = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.3,
    max_retries=2
)

os.environ["LANGCHAIN_API_KEY"]=os.getenv("LANGCHAIN_API_KEY")
os.environ["LANGCHAIN_PROJECT"]=os.getenv("LANGCHAIN_PROJECT")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

message=[
    SystemMessage(content="Translate the following from English to Hinglish"),
    HumanMessage(content="Hello How are you?")   
]

result = parser.invoke(model.invoke(message))

result

# parser.invoke(result)

```

```python
chain = model | parser

chain.invoke(message)
```

```python
#Prompt Templates
from langchain_core.prompts import ChatPromptTemplate

Generic_template = "Translate the following in {Language}:"

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", Generic_template),("user","{Text}")
    ]
)

prompt.invoke({"Language":"Hinglish","Text":"Why are you waiting?"})
```

```python
chain = prompt|model|parser

chain.invoke({"Language":"Hinglish","Text":"Lets go to Varanasi"})
```
