```python
import os
from dotenv import load_dotenv

from langchain_groq import ChatGroq

load_dotenv()

os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")

model = ChatGroq(
    model="llama-3.1-8b-instant",
    temperature=0.0,
    max_retries=2
)
```

```python
from langchain_core.messages import HumanMessage

model.invoke([HumanMessage(content = "Hi i am learning to be AI engineer")])
```

```python
from langchain_core.messages import AIMessage

model.invoke(
    [
        HumanMessage(content = "Hi i am learning to be AI engineer"),
        AIMessage(content = "That\'s exciting. Being an AI engineer can be a fascinating and rewarding career. It involves designing, developing, and deploying artificial intelligence and machine learning models to solve complex problems."),
        HumanMessage(content="What is it i want to be")
    ]
)
```

```python
##message History

from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store ={}

def get_session_history(session_id:str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id]=ChatMessageHistory()
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model, get_session_history)

```

```python
config = {"configurable":{"session_id":"chat1"}}
```

```python
response = with_message_history.invoke(
    [HumanMessage(content = "Hi i am learning to be AI engineer and My name is Krish")],
        config = config
        )

response.text
```

```python
response = with_message_history.invoke(
    [HumanMessage(content = "What is my name?")],
        config = config
        )
response.text
```

```python
config1 = {"configurable":{"session_id":"chat2"}}

response = with_message_history.invoke(
    [HumanMessage(content = "My name is Shiva")],
        config = config1
        )
response.text
```

```python
config1 = {"configurable":{"session_id":"chat2"}}

response = with_message_history.invoke(
    [HumanMessage(content = "My name is ?")],
        config = config1
        )
response.text
```

```python
from langchain_core.prompts import ChatPromptTemplate,MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are helpful assistant. Answer all the question to the best of your ability"),
        MessagesPlaceholder(variable_name = "messages")
    ]
)

chain = prompt|model
```

```python
chain.invoke({"messages":[HumanMessage(content="Ram Ram Ji !")]})
```

```python
with_message_history = RunnableWithMessageHistory(chain, get_session_history)

config = {"configurable":{"session_id":"chat3"}}

response = with_message_history.invoke(
    [HumanMessage(content="what i said earlier?")],
    config=config
)

response
```

```python
#Add More Coplexity

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","You are helpful assistant. Answer all the question to the best of your ability in {language}"),
        MessagesPlaceholder(variable_name = "messages")
    ]
)

chain = prompt|model
```

```python
response = chain.invoke({"messages":[HumanMessage(content="Jai Shiv Shankar")],"language":"Hindi"})
response.content
```

```python
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)
```

```python
config = {"configurable":{"session_id":"chat4"}}
response = with_message_history.invoke(
    {'messages':[HumanMessage(content= "Hi I'm Shiv")],"language":"Hindi"},
    config=config
)
response.content
```

```python
response = with_message_history.invoke(
    {'messages':[HumanMessage(content= "What's my name?")],"language":"Hindi"},
    config=config
)
response.content
```
