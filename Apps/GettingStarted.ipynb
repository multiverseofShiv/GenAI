```python
#Input and get Response


result = model.invoke("What is Generative Ai")
print(result.content)
```

```python
#chatprompt template 

from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages(
    [
        ("system","you are an expert AI Engineer. Provide me answers based on questions"),
        ("user","{input}")
        
    ]
)

prompt
```

```python
chain = prompt|model 

response = chain.invoke({"input":"Tell be about the pathway to become AI Engineer"})

print(response)
```

```python
# Stroutput parser
from langchain_core.output_parsers import StrOutputParser

OutputParser = StrOutputParser()

chain = prompt|model|OutputParser


response = chain.invoke({"input":"Tell be about the pathway to become AI Engineer"})

print(response)


```
